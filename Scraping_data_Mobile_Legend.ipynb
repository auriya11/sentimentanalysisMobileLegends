```python
from google_play_scraper import reviews, Sort
import pandas as pd

# Ganti dengan ID aplikasi yang ingin di-scrape
app_id = "com.mobile.legends"

# Scraping ulasan
reviews_list, _ = reviews(
    app_id,
    lang="id",  # Bahasa Indonesia
    country="id",  # Negara Indonesia
    sort=Sort.MOST_RELEVANT,  # Urutkan berdasarkan yang terbaru
    count=10000  # Jumlah ulasan yang ingin diambil
)

```

```python
# Menyimpan ulasan dalam file CSV
import csv

with open('mobile_legends_reviews1.csv', mode='w', newline='', encoding='utf-8') as file:
    writer = csv.writer(file)
    writer.writerow(['Review'])  # Menulis header kolom
    for review in reviews_list:
        writer.writerow([review['content']])  # Menulis konten ulasan ke dalam file CSV

app_reviews_df = pd.DataFrame(reviews_list)
app_reviews_df.shape
app_reviews_df.head()
app_reviews_df.to_csv('mobile_legends_reviews1.csv', index=False, encoding="utf-8")
```
